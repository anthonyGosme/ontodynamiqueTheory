## Why Docker?

The standard Colab notebook requires a **runtime restart** after `pip install` because Python cannot hot-swap libraries already loaded in RAM. This Docker image has everything pre-installed (Python, Lean 4, MDSINE2). **No restart needed.**

## How to use

1. Run the **Setup** cell — just a `docker pull`, takes ~2 min
2. Run **any section** independently, or run **all tests** at once
3. All outputs are saved to `output/`

## Setup — Pull Image & Clone Repo

```python
%%time
import os

# Pull pre-built image from Docker Hub (everything installed: Python, Lean 4, MDSINE2)
!docker pull anthonygosme/ontodynamique:latest

# Clone repository (for data files and to receive output)
if not os.path.exists('ontodynamiqueTheory'):
    !git clone https://github.com/anthonyGosme/ontodynamiqueTheory.git
    print('Repository cloned.')
else:
    print('Repository already exists, pulling latest...')
    !cd ontodynamiqueTheory && git pull

!mkdir -p ontodynamiqueTheory/output
print('\n--- Ready! No runtime restart needed. ---')
```

## View Generated Outputs

```python
import os
from IPython.display import Image, display

output_dir = 'ontodynamiqueTheory/output'
if os.path.exists(output_dir):
    for f in sorted(os.listdir(output_dir)):
        path = os.path.join(output_dir, f)
        print(f'\n--- {f} ---')
        if f.endswith('.png'):
            display(Image(filename=path, width=600))
        elif f.endswith(('.json', '.csv', '.txt')):
            with open(path) as fh:
                print(fh.read()[:2000])
else:
    print('No output directory found. Run the tests first.')
```
